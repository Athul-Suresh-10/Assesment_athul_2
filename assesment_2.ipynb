```python
#QN 13

import pandas as pd
import numpy as np

df = pd.read_csv("student_performance.csv")
df["Python_Score"] = [95, 70, 88, 55, 92, 78]
df["SQL_Score"] = [90, 75, 85, 60, 94, 72]
df["Pandas_Score"] = [88, 72, 91, 58, 89, 75]
df["NumPy_Score"] = [92, 68, 87, 62, 90, 74]

df["Average_Technical_Score"] = (
    df["Python_Score"]
    + df["SQL_Score"]
    + df["Pandas_Score"]
    + df["NumPy_Score"]
) / 4

conditions = [
    df["Average_Technical_Score"] >= 90,
    df["Average_Technical_Score"] >= 75,
    df["Average_Technical_Score"] >= 60
]

choices = [
    "Excellent",
    "Good",
    "Average"
]

df["Performance"] = np.select(
    conditions,
    choices,
    default="Needs Improvement"
)
print(df)
```

```python
#QN 14

import pandas as pd

df = pd.read_csv("student_performance.csv")
df["Final_Score"] = df[["Math_Score", "Reading_Score", "Writing_Score"]].mean(axis=1)
df["Department"] = ["Data Science", "Computer Science", "Data Science", "Physics", "Data Science", "Chemistry"]

department_summary = df.groupby("Department")["Final_Score"].agg(
    Number_of_Students="count",
    Average_Final_Score="mean",
    Maximum_Final_Score="max",
    Minimum_Final_Score="min"
).reset_index()

department_summary = department_summary.sort_values(
    by="Average_Final_Score",
    ascending=False
)
print(department_summary)
```

```python
#QN 15

import pandas as pd
import numpy as np


df = pd.read_csv("student_performance.csv")


df["Final_Score"] = df[["Math_Score", "Reading_Score", "Writing_Score"]].mean(axis=1)


scores = df["Final_Score"].to_numpy()


mean = np.mean(scores)
median = np.median(scores)
std_dev = np.std(scores)
percentile_25 = np.percentile(scores, 25)
percentile_75 = np.percentile(scores, 75)

print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std_dev)
print("25th Percentile:", percentile_25)
print("75th Percentile:", percentile_75)


above_mean = df[df["Final_Score"] > mean]

print("\nStudents with Final_Score above the overall mean:")
print(above_mean)
```

```python
#QN 16

import pandas as pd
import numpy as np


df = pd.read_csv("student_performance.csv")



print("Missing Values:")
print(df.isnull().sum())


numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("\nData after handling missing values:")
print(df)


df["Python_Score"] = [95, 70, 88, 55, 92, 78]
df["SQL_Score"] = [90, 75, 85, 60, 94, 72]
df["Pandas_Score"] = [88, 72, 91, 58, 89, 75]
df["NumPy_Score"] = [92, 68, 87, 62, 90, 74]
df["Attendance"] = [95, 70, 85, 60, 92, 78]
df["Department"] = ["Data Science", "Computer Science", "Data Science", "Physics", "Data Science", "Chemistry"]





df["Average_Score"] = df[
    ["Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score"]
].mean(axis=1)


df["Pass_Status"] = np.where(
    df["Average_Score"] >= 50,
    "Pass",
    "Fail"
)

print("\nData after Feature Engineering:")
print(df)



top_5 = df.sort_values(
    "Average_Score",
    ascending=False
).head(5)

print("\nTop 5 Students:")
print(top_5)



department_scores = df.groupby("Department")["Average_Score"].mean()

print("\nAverage Score by Department:")
print(department_scores)

print("\nDepartment with Highest Average Score:")
print(department_scores.idxmax())


attendance = df.groupby("Department")["Attendance"].mean()

print("\nAverage Attendance by Department:")
print(attendance)



print("\nNumber of Students Passed and Failed:")
print(df["Pass_Status"].value_counts())


summary = df.groupby("Department").agg(
    Total_Students=("Department", "count"),
    Average_Score=("Average_Score", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
    Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
).reset_index()

print("\nSummary DataFrame:")
print(summary)
```
